# what is variables in python?
Variables in Python are used to store data values. A variable is essentially a name that refers to a value in memory. You can think of it as a container that holds information that can be used and manipulated throughout your program.
## example of variables in python
```python
name = "Alice"
age = 25
height = 5.6
is_student = True
```
## types of variables in python
Python has several built-in data types that can be used as variables, including:
- Integer (int): Whole numbers, e.g., `age = 25`
- Float (float): Decimal numbers, e.g., `height = 5.6`
- String (str): Text, e.g., `name = "Alice"`
- Boolean (bool): True or False values, e.g., `is_student = True`
- List (list): Ordered collection of items, e.g., `fruits = ["apple", "banana", "cherry"]`
- Tuple (tuple): Immutable ordered collection of items, e.g., `coordinates = (10, 20)`
- Dictionary (dict): Key-value pairs, e.g., `person = {"name": "Alice", "age": 25}`
## variable naming rules in python
When naming variables in Python, you should follow these rules:
- Variable names must start with a letter (a-z, A-Z) or an underscore (_).
- The rest of the variable name can contain letters, numbers (0-9), and underscores
- Variable names are case-sensitive (e.g., `age` and `Age` are different variables).
- Avoid using Python reserved keywords (e.g., `if`, `for`, `while`, etc.) as variable names.
## conclusion
Variables are fundamental to programming in Python, allowing you to store and manipulate data effectively. Understanding their types and naming conventions is crucial for writing clean and efficient code.


````python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
year_of_birth = 2025 - age
print(f"Hello, {name}! You were born in {year_of_birth}.")
````

## conditional statements in python
Conditional statements in Python are used to perform different actions based on whether a certain condition is true or false. The primary conditional statements in Python are `if`, `elif`, and `else`.
### example of conditional statements in python
```python
age = 18
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```
### types of conditional statements in python
1. `if` statement: Used to test a specific condition. If the condition is true, the block of code inside the `if` statement is executed.
```python
if condition:
    # code block
```
2. `elif` statement: Short for "else if," it allows you to check multiple conditions. If the first `if` condition is false, the program checks the `elif` condition.
```python
if condition1:
    # code block 1
elif condition2:
    # code block 2
```
3. `else` statement: Used to execute a block of code when none of the previous conditions are true.
```python
if condition1:
    # code block 1
elif condition2:
    # code block 2
else:
    # code block 3
```
### example with multiple conditions
```python
score = 85
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```
### conclusion
Conditional statements are essential for controlling the flow of a program based on different conditions. They allow you to make decisions and execute specific blocks of code accordingly.

````python
if age < 18:
    print("You are a minor.")
else:
    print("You are an adult.")
    
````

# looping statements in python
Looping statements in Python are used to execute a block of code repeatedly as long as a certain condition is met. The primary looping statements in Python are `for` and `while` loops.
### example of looping statements in python
```python
# Using a for loop to iterate over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```
### types of looping statements in python
1. `for` loop: Used to iterate over a sequence (like a list, tuple, or string) or other iterable objects. The loop runs for each item in the sequence.
```python   
for item in sequence:
    # code block
```
2. `while` loop: Used to execute a block of code as long as a specified condition is true.
```python
while condition:
    # code block
```
### example of while loop
```python
count = 0
while count < 5:
    print(count)
    count += 1
```
### conclusion
Looping statements are crucial for performing repetitive tasks in programming. They allow you to iterate over data structures and execute code multiple times based on conditions.


````python
n = 5
for i in range(1, 11):
    print(f"{n} x {i} = {n * i}")
````

# functions in python
Functions in Python are reusable blocks of code that perform a specific task. They help organize code, improve readability, and allow for code reuse. Functions are defined using the `def` keyword, followed by the function name and parentheses that may include parameters.
### example of functions in python
```python
def greet(name):
    print(f"Hello, {name}!")    
greet("Alice")  # Output: Hello, Alice!
``` 
### types of functions in python
1. Built-in functions: Python provides many built-in functions like `print()`, `len()`, `type()`, etc., that can be used directly without any definition.
2. User-defined functions: These are functions that you create using the `def` keyword to perform specific tasks.
```python   
def add(a, b):
    return a + b        
result = add(5, 3)  # result is 8
``` 
3. Lambda functions: These are small anonymous functions defined using the `lambda` keyword. They can take any number of arguments but can only have one expression.
```python   
square = lambda x: x * x        
result = square(4)  # result is 16
```
### example of function with multiple parameters
```python
def multiply(x, y):
    return x * y
result = multiply(4, 5)  # result is 20
```
### conclusion
Functions are essential for structuring code in Python. They allow you to encapsulate logic, making your code more modular and easier to maintain.

````python
def factorial(num):
    if num == 0 or num == 1:
        return 1
    else:
        return num * factorial(num - 1)
print(f"Factorial of 5 is {factorial(5)}")
````

````python
multi = lambda x, y: x * y
print(f"Multiplication of 4 and 5 is {multi(4, 5)}")
````

# List in Python
A list in Python is a built-in data type that allows you to store an ordered collection of items. Lists are mutable, meaning you can change their content after they are created. They can contain items of different data types, including integers, strings, and even other lists.
### example of list in python
```python
# Creating a list
fruits = ["apple", "banana", "cherry"]
# Accessing list items
print(fruits[0])  # Output: apple
# Modifying list items
fruits[1] = "blueberry"
print(fruits)  # Output: ['apple', 'blueberry', 'cherry']
```
### list operations in python
1. Adding items: You can add items to a list using the `append()` method or
the `insert()` method.
```python
fruits.append("orange")  # Adds 'orange' to the end of the list
fruits.insert(1, "kiwi")  # Inserts 'kiwi' at index
2. Removing items: You can remove items using the `remove()` method or the `pop()` method.
```python
fruits.remove("banana")  # Removes 'banana' from the list
fruits.pop(2)  # Removes the item at index 2
3. Slicing: You can access a subset of the list using slicing.
```python   
subset = fruits[1:3]  # Gets items from index 1 to 2
4. Iterating: You can loop through the items in a list using a `for` loop.
```python
for fruit in fruits:
    print(fruit)    
### conclusion
Lists are versatile and powerful data structures in Python that allow you to store and manipulate collections of items easily. They are widely used in various applications due to their flexibility and ease of use.
# Tuples in Python
A tuple in Python is a built-in data type that allows you to store an ordered collection of items. Unlike lists, tuples are immutable, meaning that once they are created, their content cannot be changed. Tuples can contain items of different data types, including integers, strings, and even other tuples.
### example of tuple in python
```python
# Creating a tuple
coordinates = (10, 20, 30)
# Accessing tuple items
print(coordinates[0])  # Output: 10
# Tuples are immutable, so you cannot modify their items
```
### tuple operations in python
1. Creating single-item tuples: To create a tuple with a single item, you need to
include a comma after the item.
```python
single_item_tuple = (5,)
2. Unpacking: You can unpack the items of a tuple into separate variables.
```python
x, y, z = coordinates  # x=10, y=20, z=30
3. Slicing: You can access a subset of the tuple using slicing.
```python
subset = coordinates[1:3]  # Gets items from index 1 to 2
4. Iterating: You can loop through the items in a tuple using a `for` loop.
```python
for coord in coordinates:
    print(coord)    
# conclusion
Tuples are useful for storing collections of items that should not be modified. They provide a way
to group related data together while ensuring that the data remains unchanged throughout the program.
# Dictionary in Python
A dictionary in Python is a built-in data type that allows you to store key-value pairs. Dictionaries are mutable, meaning you can change their content after they are created. They are unordered collections, which means that the items do not have a defined order.
### example of dictionary in python
```python   
# Creating a dictionary
person = {"name": "Alice", "age": 25, "city": "New York"}
# Accessing dictionary items
print(person["name"])  # Output: Alice
# Modifying dictionary items
person["age"] = 26
print(person)  # Output: {'name': 'Alice', 'age': 26, 'city': 'New York'}
```
### dictionary operations in python
1. Adding items: You can add new key-value pairs to a dictionary by assigning a value to a new key.
```python
person["job"] = "Engineer"  # Adds a new key-value pair
2. Removing items: You can remove items using the `del` statement or the `pop()` method.
```python
del person["city"]  # Removes the key 'city'
person.pop("age")  # Removes the key 'age' and returns its value
3. Iterating: You can loop through the keys, values, or key-value pairs in a dictionary using a `for` loop.
```python
for key in person:
    print(key, person[key])  # Prints each key and its corresponding value
for key, value in person.items():
    print(key, value)  # Prints each key-value pair
### conclusion
Dictionaries are powerful data structures in Python that allow you to store and manipulate data using key-value pairs. They are widely used for various applications due to their flexibility and ease of use.

# OOP in Python
Object-Oriented Programming (OOP) in Python is a programming paradigm that uses "objects"
to represent data and functionality. OOP allows for the organization of code into reusable and modular components, making it easier to manage and maintain complex software systems. The four main principles of OOP are encapsulation, inheritance, polymorphism, and abstraction.
### example of OOP in python
```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def bark(self):
        return f"{self.name} says Woof!"
my_dog = Dog("Buddy", 3)
print(my_dog.bark())  # Output: Buddy says Woof!
```

# classes and objects in python
In Python, a class is a blueprint for creating objects. An object is an instance of a class. Classes encapsulate data and functionality together, allowing you to create multiple objects with similar properties and behaviors.

# what is class attributes and instance attributes in python
In Python, class attributes and instance attributes are two types of attributes that can be defined within a class.
### class attributes
Class attributes are shared by all instances of a class. They are defined within the class but outside of any instance methods. Class attributes are accessed using the class name or through an instance of the class.
```python   
class Dog:
    species = "Canis familiaris"  # Class attribute
    def __init__(self, name, age):
        self.name = name  # Instance attribute
        self.age = age    # Instance attribute
``` 
### instance attributes
Instance attributes are specific to each instance of a class. They are defined within the `__init__` method and are accessed using the instance of the class.
```python
my_dog = Dog("Buddy", 3)
print(my_dog.name)  # Output: Buddy (instance attribute)
print(Dog.species)  # Output: Canis familiaris (class attribute)
```

````python
class Person:
    id = 122345 # class variable
    def __init__(self, name, age): # constructor
        self.name = name # instance variable
        self.age = age
    def greet(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."
print(f"Person ID: {Person.id}")  # accessing class variable using class name
p1= Person("Alice", 30)
print(p1.greet())  # access instance variables using object of class
print(p1.__dict__)
````

# properties in python
In Python, properties are a way to manage the access to instance attributes. They allow you to define methods that can be accessed like attributes, providing a way to add logic to getting, setting, or deleting an attribute's value.
### example of properties in python
```python
class Circle:
    def __init__(self, radius):
        self._radius = radius  # Private instance attribute
    @property
    def radius(self):
        return self._radius
    @radius.setter
    def radius(self, value):
        if value < 0:
            raise ValueError("Radius cannot be negative.")
        self._radius = value
    @property
    def area(self):
        import math
        return math.pi * (self._radius ** 2)
circle = Circle(5)
print(circle.radius)  # Output: 5
print(circle.area)    # Output: 78.53981633974483
circle.radius = 10
print(circle.radius)  # Output: 10
```
