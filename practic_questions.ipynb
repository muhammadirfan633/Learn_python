Create a function normalize_email(email) → lower case + strip spaces

```python
def normalize_mail(email):
    email = email.strip().lower()
    return email
user_email = input("Enter your email: ")
normalized_email = normalize_mail(user_email)
print(f"Normalized email: {normalized_email}")
```

Create a function slugify(title) → convert to URL slug ("My Post Title" → my-post-title)

```python
def slugify(text):
    text = text.strip().lower().replace(" ", "-")
    return text
user_input = input("Enter text to slugify: ")
slugified_text = slugify(user_input)
print(f"Slugified text: {slugified_text}")
```

Create a function generate_username(name) → remove spaces + add random number

```python

import random
def generate_username(full_name):
    number = random.randint(100, 999)
    username = full_name.strip().lower().replace(" ", "")+ str(number)
    return username
user_name = input("Enter your full name: ")
generated_username = generate_username(user_name)
print(f"Generated username: {generated_username}")
```

Create a function is_strong_password(password) (min 8 chars, 1 digit, 1 uppercase)

```python
def is_strong_password(password):
    if (len(password) >= 8 and
        any(char.isdigit() for char in password) and
        any(char.isupper() for char in password) and
        any(char.islower() for char in password) and
        any(char in "!@#$%^&*()-_+=" for char in password)):
        return True
    return False
user_password = input("Enter your password: ")
if is_strong_password(user_password):
    print("Strong password")
else:
    print("Weak password")
```

Sort products by price (low to high)

```python
product = [{"name": "Laptop", "price": 1200, "quantity": 5}, {"name": "Mouse", "price": 25, "quantity": 10}, {"name": "Keyboard", "price": 45, "quantity": 7}]
sort = sorted(product, key=lambda x: x["price"])
print(f"Products sorted by price: {sort}")
```

Paginate list manually (page=2, page_size=5)

```python
page = int(input("Enter page number: "))
per_page = int(input("Enter items per page: "))
def paginate(items, page, per_page):
    start = (page - 1) * per_page
    end = start + per_page
    return items[start:end]
items = list(range(1, 101))  # Example list of 100 items
paginated_items = paginate(items, page, per_page)
print(f"Items on page {page}: {paginated_items}")
```

Create a list of dictionaries as a fake database and perform:

```python
users = [{"name":"Ali","active":True}, {"name":"Sara","active":False}, {"name":"John","active":True}, {"name":"Lina","active":False}]
print("create a new user: enter 1")
print("read users: enter 2")
print("update user: enter 3")
print("delete user: enter 4")
print("exit: enter 5")
run = True
while run:
    choice = input("Enter your choice: ")
    if choice == "1":
        name = input("Enter name: ")
        active = input("Is active (True/False): ").strip().lower() == "true"
        users.append({"name": name, "active": active})
        print(f"User {name} added.")
    elif choice == "2":
        name = input("Enter name to search: ")
        found_users = [user for user in users if user["name"] == name]
        print(f"Found users: {found_users}")
    elif choice == "3":
        name = input("Enter name to update: ")
        for user in users:
            if user["name"] == name:
                new_name = input("Enter new name: ")
                user["name"] = new_name
                print(f"User {name} updated to {new_name}.")
                break
        else:
            print(f"User {name} not found.")
    elif choice == "4":
        name = input("Enter name to delete a user: ")
        if name in [user["name"] for user in users]:
            users = [user for user in users if user["name"] != name]
            print(f"User {name} deleted.")
        else:
            print(f"User {name} not found.")
    elif choice == "5":
        print("Exiting program.")
        run = False
    
```
