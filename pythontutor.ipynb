# list of data types

**List** in Python is a built-in data type that is used to store an ordered collection of items. It is mutable, meaning that you can modify the contents of a list after it has been created. Lists can contain elements of different data types, including numbers, strings, and even other lists.

example:


````python
mylist = [1, 2, 3, 4, 5]
print(mylist[0])  # Output: 1
````

````python
# print list in reverse order
print(mylist[-1::-1])  # Output: [5, 4, 3, 2, 1]
````

* **indexing:** Lists are indexed, meaning that each element in a list has a specific position or index. The first element is at index 0, the second element is at index 1, and so on. You can access individual elements of a list using their index.

* **Negative indexing:** You can also use negative indexing to access elements from the end of the list. The last element is at index -1, the second-to-last element is at index -2, and so on.

* **slicing:** You can use slicing to access a range of elements in a list. The syntax for slicing is 
**list[start:stop:step]**, where start is the index of the first element, stop is the index of the last element (exclusive), and step is the interval between elements.

* **modifying lists:** Since lists are mutable, you can modify their contents by assigning new values to specific indices or by using methods like append(), insert(), remove(), and pop().

````python
L = [1, 2, 3, 4, 5]
L[0] = 10
print(L)  # Output: [10, 2, 3, 4,
````

## Recursion
Recursion is a programming technique where a function calls itself in order to solve a problem. A recursive function typically has a base case that stops the recursion and a recursive case that breaks the problem into smaller subproblems. Recursion can be a powerful tool for solving problems that can be broken down into smaller, similar problems, such as tree traversal, factorial calculation, and Fibonacci sequence generation. However, it is important to ensure that the base case is reached to avoid infinite recursion and potential stack overflow errors.


example:

```python
def recursive_factorial(n):
    if n == 0:
        return 1  # Base case: factorial of 0 is 1
    else:
        return n * recursive_factorial(n - 1)  # Recursive case: n! = n * (n-1)!
    
print(recursive_factorial(5))  # Output: 120

````python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
print(factorial(5))  # Output: 120
````

* **Inheritance** allows a new class to inherit properties and behaviors from an existing class. 
* **Polymorphism** enables objects of different classes to be treated as objects of a common superclass. 
* **Abstraction** involves hiding the complex implementation details and exposing only the necessary features of an object.

* ## Polymorphism
Polymorphism is a fundamental concept in object-oriented programming that allows objects of different classes to be treated as objects of a common superclass. It enables a single interface to represent different underlying data types or classes. Polymorphism can be achieved through method overriding, where a subclass provides a specific implementation of a method that is already defined in its superclass. This allows for dynamic method dispatch, where the appropriate method is called based on the actual type of the object at runtime. Polymorphism promotes code reusability and flexibility, as it allows for the use of a common interface to work with objects of different types without needing to know their specific classes.


**Types of Polymorphism:**
1. Compile-time Polymorphism (Method Overloading): This type of polymorphism occurs when multiple methods in the same class have the same name but different parameters. The method to be called is determined at compile time based on the number and type of arguments passed.
2. Runtime Polymorphism (Method Overriding): This type of polymorphism occurs when a subclass provides a specific implementation of a method that is already defined in its superclass. The method to be called is determined at runtime based on the actual type of the object.

Compile time polymorphism (method overloading) is not directly supported in Python, as it does not allow multiple methods with the same name but different parameters. However, you can achieve a similar effect using default arguments or variable-length arguments. For example:

```python   
class MyClass:
    def multiply(self, a=2, b=1,*args):
        result = a * b
        for num in args:
            result *= num
        return result
m= MyClass()
print(m.multiply(5, 3))  # Output: 15
print(m.multiply(5, 3, 2))  # Output: 30
print(m.multiply(5))  # Output: 10


````python
class myclass:
    def multiply(self, a=2, b=3, *args):
        result = a * b
        for num in args:
            result *= num
        return result
m= myclass()
print(m.multiply(5,6))  # Output: 6
print(m.multiply(2, 3, 4))  # Output: 24
````

## Abstraction
Abstraction is the principle of hiding the complex implementation details and exposing only the necessary features of an object. It allows for the creation of abstract classes and interfaces that define a common interface for a group of related classes, while hiding the specific implementation details of each class. Abstraction promotes code reusability and maintainability by providing a clear separation between the interface and implementation of an object. In Python, abstraction can be achieved using abstract base classes (ABCs) and the abc module, which allows you to define abstract methods that must be implemented by any subclass. For example:
```python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass    
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius ** 2  
circle = Circle(5)
print(circle.area())  # Output: 78.5


````python
from abc import ABC, abstractmethod
class greeding(ABC):
    @abstractmethod
    def greet(self):
        print("hello")
class english_greeting(greeding):
    def greet(self):
        return "Hello! khan"
e = english_greeting()
print(e.greet())  # Output: Hello! khan
````

# Supper() function
The super() function in Python is used to call a method from a parent class in a child class. ```python

# async and await
In Python, async and await are keywords used to define and work with asynchronous programming. The await keyword is used to pause the execution of an asynchronous function until a certain condition is met or until another asynchronous function completes. deffirence between async and sync functions:

2. Blocking: Synchronous functions block the execution of other code until they complete, while asynchronous functions do not block and can allow other tasks to run while waiting for a certain condition or event to occur.

# What is Multithreading and Multiprocessing in Python
Multithreading and multiprocessing are two approaches to achieving concurrent execution in Python, allowing you to perform multiple tasks simultaneously.
1. Multithreading: Multithreading is a technique where multiple threads of execution run concurrently within a single process. Each thread can perform a different task, and they share the same memory space. This allows for efficient use of resources and can improve the performance of I/O-bound tasks, such as network requests or file handling. However, due to Python's Global Interpreter Lock (GIL), multithreading may not be effective for CPU-bound tasks that require intensive computation.
2. Multiprocessing: Multiprocessing is a technique where multiple processes run concurrently, each with its own memory space. This allows for true parallelism and can improve the performance of CPU-bound tasks, as each process can utilize multiple CPU cores. However, multiprocessing can be more resource-intensive than multithreading, as it involves creating separate processes and managing inter-process communication.
In Python, you can use the threading module for multithreading and the multiprocessing module for multiprocessing. For example:
```python
import threading
import multiprocessing
# Example of multithreading
def thread_function(name):
    print(f"Thread {name} is running.")
thread1 = threading.Thread(target=thread_function, args=("A",))
thread2 = threading.Thread(target=thread_function, args=("B",))
thread1.start()
thread2.start()
# Example of multiprocessing
def process_function(name):
    print(f"Process {name} is running.")
process1 = multiprocessing.Process(target=process_function, args=("A",))
process2 = multiprocessing.Process(target=process_function, args=("B",))
process1.start()
process2.start()
